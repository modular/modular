```python
%%mojo
fn incr_int(inout arg: Int):
    arg += 1  
```

```python
import unittest
import sys

class TestMojoMagicCell(unittest.TestCase):
    def test_calling_toplevel_mojo_func(self):
        mojo_int = Int()
        self.assertEqual(type(mojo_int), Int)
        self.assertEqual(repr(mojo_int), "0")
        incr_int(mojo_int)
        self.assertEqual(repr(mojo_int), "1")
        
result = unittest.main(argv=[''], verbosity=2, exit=False)
```

```python
%%mojo verbose=1
fn square(inout arg: Int):
  arg = arg * arg
```

```python
val = Int()
incr_int(val)
incr_int(val)
for i in range(6):
    print(repr(val))
    square(val)
```
